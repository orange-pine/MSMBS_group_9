```python
import numpy as np
import pandas as pd
from scipy.integrate import odeint
import matplotlib.pyplot as plt
import seaborn as sns
import cobra
```

```python
model = cobra.io.load_json_model('e_coli_core-1.json')

```

```python
print("Number of metabolites:", len(model.metabolites))
print("Number of reactions:", len(model.reactions))
print("Number of genes:", len(model.genes))
```

```python
print(model.metabolites)
print(model.reactions)
print(model.genes)
```

```python
for key, value in model.compartments.items():
    print(key, "=", value)
```

```python
for reaction in model.exchanges:
    print(reaction.id, ":", reaction.name)
```

```python
print("Number of reactions:", len(model.reactions))
print("Number of genes:", len(model.genes))
```

```python
pfk = model.reactions.get_by_id("PFK")
print(pfk.genes)
```

```python
print("ID:", pfk.id)
print("Name:", pfk.name)
print("Reaction:", pfk.reaction)
print("Reversible:", pfk.reversibility)
print("Genes:", pfk.genes)
```

```python
for metabolite, coefficient in pfk.metabolites.items():
    print(metabolite.id, coefficient)
```
